```python
import zipfile
import os

def extract_folder_from_zip(zip_path, target_folder, extract_to):
    """
    Extrait un dossier spécifique d'une archive ZIP.

    :param zip_path: chemin vers le fichier .zip
    :param target_folder: dossier à extraire (ex: "benign/")
    :param extract_to: dossier de destination
    """
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        members = [m for m in zip_ref.namelist() if m.startswith(target_folder)]
        if not members:
            print(f"Dossier '{target_folder}' introuvable dans le zip.")
            return
        zip_ref.extractall(path=extract_to, members=members)
        print(f"Dossier '{target_folder}' extrait dans : {extract_to}")
```

```python
zip_file = "../data/BHI.zip"            # ton fichier .zip
folder_name = "IDC_regular_ps50_idx5/"                   # nom du dossier à extraire
destination = "../data"     # destination des fichiers extraits

extract_folder_from_zip(zip_file, folder_name, destination)
```
